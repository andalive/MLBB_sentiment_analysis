Notebook ini digunakan untuk memperoleh dataset ulasan pengguna Mobile Legends: Bang Bang (MLBB) dari Google Play Store sebagai data mentah untuk digunakan pada tahap analisis sentimen.

**Pemilihan Sumber Data**

**Google Play Store** dipilih sebagai sumber data karena menyediakan ulasan berisi pengalaman, penilaian, keluahan, hingga apresiasi pengguna terhadap aplikasi yang digunakan. Lalu, terdapat skor penilaian yang dapat dimanfaatkan sebagai salah satu dasar pelebelan sentimen tahap analisis.

**Pemilihan Aplikasi**

**MLBB** dipilih berdasar pada karakteristik apalikasi yang memiliki basis pengguna dan jumlah ulasan cukup besar. Berbagai ulasan terkait pengalaman pengguna dapat memberikan peluang analisis sentimen yang beragam. Selain itu, MLBB juga membuat proyek ini lebih dekat dengan konteks yang familiar bagi penulis.

## 01. IMPORT LIBRARY

```python
from google_play_scraper import Sort, reviews_all
import pandas as pd
```

## 02. SCRAPING DATASET

Proses scraping data ulasan dilakukan dengan `google-play-scraper`.</br>
**Source**: [JoMingyu/google-play-scraper](https://github.com/JoMingyu/google-play-scraper)


```python
# Mengambil semua ulasan dari aplikasi tertentu di Google Play Store
app_id = 'com.mobile.legends'

scraper_results = reviews_all(
    app_id,
    sleep_milliseconds=100,  # defaults 0
    lang='id',               # dafaults 'en'
    country='id',            # defaults 'us'
    sort=Sort.MOST_RELEVANT, # defaults Sort.MOST_RELEVANT
    filter_score_with=None   # defaults None(means all score)
)

print(f"Total reviews diambil: {len(scraper_results)}")
```

```python
# Konversi hasil scraping menjadi DataFrame
df = pd.DataFrame(scraper_results)
df.head()
```

```python
# Deskripsi DataFrame
print(df.shape)
df.info()

# Menyimpan hasil ulasan ke dalam file CSV
CSV_file = 'MLBB_scraped_reviews.csv'

df.to_csv(CSV_file, index=False, encoding='utf-8-sig')
print(f"\nHasil CSV disimpan di {CSV_file}")
```
